```python
import pandas as pd
from google.colab import files

# Upload the Excel file
uploaded = files.upload()

# Get the uploaded filename
filename = list(uploaded.keys())[0]

# Read Excel file into a dataframe
df = pd.read_excel(filename)

# Display the first five rows
df.head(10)

# Column names
print(df.columns)

# Data types
print(df.dtypes)
```

```python
# Identify columns with 'object' (categorical) data type
object_cols = df.select_dtypes(include=['object']).columns

print(f"Found {len(object_cols)} object-type columns: {list(object_cols)}\n")

# Loop through each object column and print its unique values
for col in object_cols:
    unique_vals = df[col].unique()
    print(f"Unique values in '{col}' ({len(unique_vals)} unique):")
    print(unique_vals)
    print("-" * 40)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter the data for the 'all' situation
all_situation_df = df[df['situation'] == 'all']

# Create the plot
plt.figure(figsize=(10, 6))
sns.histplot(data=all_situation_df, x='xGoals', kde=True, color='skyblue', bins=20)

# Customize labels and title
plt.title("Distribution of 'xGoals' (Situation: 'all')", fontsize=14)
plt.xlabel("Expected Goals (xGoals)", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Show the visualization
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot comparing xGoals vs goals
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='xGoals', y='goals', hue='situation', alpha=0.7)

# Add a diagonal line representing xGoals = goals for reference
max_val = max(df['xGoals'].max(), df['goals'].max())
plt.plot([0, max_val], [0, max_val], color='red', linestyle='--', label='Perfect Agreement (xGoals = Goals)')

# Customize labels, title, and legend
plt.title("Comparison of Expected Goals (xGoals) vs Actual Goals", fontsize=14)
plt.xlabel("Expected Goals (xGoals)", fontsize=12)
plt.ylabel("Actual Goals", fontsize=12)
plt.legend(title='Situation')
plt.grid(True, linestyle='--', alpha=0.5)

# Show plot
plt.show()
```

```python
# Filter for the 'all' situation to have a standardized comparison per goalie
goalie_perf = df[df['situation'] == 'all'].copy()

# Calculate the difference: actual goals minus expected goals
# A negative value means the goalie allowed fewer goals than expected (better performance)
goalie_perf['goals_minus_xGoals'] = goalie_perf['goals'] - goalie_perf['xGoals']

# Sort by this metric to get the lowest values (best performance)
top_10_goalies = goalie_perf.sort_values(by='goals_minus_xGoals').head(10)

# Display the relevant columns for the top 5 performers
display(top_10_goalies[['name', 'team', 'goals', 'xGoals', 'goals_minus_xGoals']])
```
